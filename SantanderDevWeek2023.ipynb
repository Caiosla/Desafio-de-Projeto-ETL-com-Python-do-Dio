# 🏦 Santander Dev Week 2023 — Versão Adaptada
### ETL com Python puro (sem IA, sem dependências externas)

Projeto baseado no desafio original da DIO.

**Fluxo ETL:**
- **Extract** → Dados dos clientes em JSON local
- **Transform** → Mensagens geradas por templates Python
- **Load** → Resultado salvo em arquivo JSON

```python
# EXTRACT — dados dos clientes (substitui o GET da API original)

users = [
    {
        "id": 1,
        "name": "Ana Souza",
        "account": {"balance": 624.12, "limit": 1000.0},
        "card": {"number": "**** **** **** 6116", "limit": 2000.0},
        "news": []
    },
    {
        "id": 2,
        "name": "Carlos Mendes",
        "account": {"balance": 1250.00, "limit": 2000.0},
        "card": {"number": "**** **** **** 3421", "limit": 3500.0},
        "news": []
    },
    {
        "id": 3,
        "name": "Beatriz Lima",
        "account": {"balance": 320.50, "limit": 500.0},
        "card": {"number": "**** **** **** 9988", "limit": 1000.0},
        "news": []
    }
]

print(f"✅ {len(users)} clientes carregados!")
for u in users:
    print(f"   → [{u['id']}] {u['name']} | Saldo: R$ {u['account']['balance']:.2f}")
```

```python
# TRANSFORM — gera mensagens por templates (substitui o ChatGPT original)
import random

def generate_news(user):
    nome = user['name'].split()[0]
    saldo = user['account']['balance']

    templates = [
        f"{nome}, seu saldo de R${saldo:.2f} pode render muito mais. Invista agora!",
        f"Que tal fazer R${saldo:.2f} trabalhar por você, {nome}? Comece a investir!",
        f"{nome}, investir é o primeiro passo para a liberdade financeira. Não espere!",
        f"R${saldo:.2f} parado não cresce, {nome}. Descubra nossas opções de investimento!"
    ]

    return random.choice(templates)

for user in users:
    news = generate_news(user)
    print(news)
    user['news'].append({
        "icon": "https://digitalinnovationone.github.io/santander-dev-week-2023-api/icons/credit.svg",
        "description": news
    })

print("\n✅ Transform concluído!")
```

```python
# LOAD — salva o resultado em JSON local (substitui o PUT da API original)
import json

def save_results(users, filename='clientes_atualizados.json'):
    with open(filename, 'w', encoding='utf-8') as f:
        json.dump(users, f, ensure_ascii=False, indent=2)
    return filename

arquivo = save_results(users)
print(f"✅ Resultado salvo em '{arquivo}'!\n")

print("📋 Resumo das mensagens geradas:")
print("-" * 60)
for user in users:
    print(f"👤 {user['name']}")
    print(f"   💬 {user['news'][0]['description']}")
    print()
```
